```python
# ==============================================================================
# STEP 1: LLM REASONING & EXPANSION (GEMMA 4 31B)
# ==============================================================================
import gc
import json
import os
import re
import pandas as pd
import torch

# Assuming llm, tokenizer, json_repair, SamplingParams are already defined/loaded
# Working directly with your exact coordinate dataset context
df_exact = exact_coordinates_airstrikes.copy()

generator_prompt = """You are an elite Intelligence Analyst specializing in weapon systems and weapon system signatures.
Read the following structured military incident packet and write a highly detailed, analytical intelligence brief.

*** MANDATORY ANALYTICAL SCRATCHPAD (<think> block) ***
Before writing your final brief, you MUST open a `<think>` block to analyze:

1. ACTOR CORRECTION & GEOPOLITICAL TIMELINE:
   - Identify the date of the event and the geographic province. Cross-reference this timestamp and geography with known military campaigns.
   - If the recorded Actor is "UNK", read the report text closely to deduce the true responsible nation-state actor (e.g., matching targeted units like IRGC, police, or naval bases with the likely attacking state).

2. TARGET INTERACTION:
   - Evaluate the specific Target Subtype (e.g., naval base, missile base, industrial facility, IRGC building).
   - Assess how the nature of this facility relates to the weapon class needed to damage it (e.g., deep bunker vs. open surface asset).

3. STRUCTURAL DAMAGE:
   - STEP A (STRUCTURAL IMPACT): Based on the text and your operational deduction, what is the level of destruction? Pick EXACTLY ONE:
     * "Total Annihilation": Entire buildings leveled, massive craters, catastrophic damage.
     * "Severe Structural": Partial building collapse, heavy fire, severe damage.
     * "Targeted / Room-Level": Specific room/floor destroyed, building remains standing.
     * "Minor / Surface": Shattered glass, open field craters, shrapnel marks, minor fires.
     * "Unknown": No damage details provided.

4. WEAPON CLASSIFICATION:
   - STEP B (WEAPON IDENTIFICATION & QUANTITIES): Identify ALL weapon types mentioned and deduce their quantities. If a specific number isn't provided (e.g., "a swarm", "several", "a barrage"), estimate a logical integer based on the context. If singular, use 1.
   - Classify EACH identified weapon into one of these strict categories:
     * "Small Kamikaze Drone" (e.g., quadcopters, small FPVs)
     * "Heavy Loitering Munition" (e.g., Shahed-136 type)
     * "Unguided Rocket" (e.g., Grad, Qassam, tactical rocket arrays)
     * "Heavy Artillery / Mortar" (e.g., 155mm shells)
     * "Air-to-Surface / Smart Bomb" (e.g., Fighter jet strikes, JDAMs, Spice glide bombs)
     * "Cruise Missile"
     * "Ballistic Missile"
     * "Unknown Generic"

*** URGENT DIRECTIVE ***
Write your brief inside `<intelligence_brief>` tags. Separate distinct targets with "EVENT [X]" headers.

CRITICAL ARRAY FORMATTING FOR FINAL OUTPUT:
EVENT 1:
- weapons_breakdown: [List EACH weapon class and its quantity, e.g., "6 Ballistic Missile, 10 Heavy Loitering Munition"]
- structural_impact: [Insert EXACT string from Step A]
"""

texts_to_process = df_exact["SIGACT"].tolist()
indices_to_process = df_exact.index.tolist()

llm_prompts = []
for idx, text in enumerate(texts_to_process):
    if not isinstance(text, str) or text.strip() == "":
        llm_prompts.append("")
    else:
        current_idx = indices_to_process[idx]
        row_data = df_exact.loc[current_idx]

        actor_val = str(row_data.get("Actor", "UNK"))
        province_val = str(row_data.get("Province", "Unknown"))
        site_type_val = str(
            row_data.get(
                "Site_Type__Nuclear__Military__Energy__etc__", "Unknown"
            )
        )
        subtype_val = str(row_data.get("SiteSubtype", "Unknown"))
        date_val = str(row_data.get("Post_Date", "Unknown"))

        structured_context = f"""
[METADATA ENVIRONMENT]
- Logged Event Date: {date_val}
- Geographic Province: {province_val}
- Categorized Actor: {actor_val}
- Primary Facility Class: {site_type_val}
- Specific Target Subtype: {subtype_val}

[RAW REPORT NARRATIVE]
{text}
"""

        messages = [
            {"role": "system", "content": generator_prompt},
            {
                "role": "user",
                "content": f"{structured_context.strip()}\n\nWrite the expanded brief.",
            },
        ]
        prompt = tokenizer.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
        llm_prompts.append(prompt)

# Sampling logic optimized for deterministic reasoning tasks via Gemma 4 31B
llm_sampling = SamplingParams(temperature=0.1, top_p=0.95, top_k=64, max_tokens=1500)

print(f"Generating Intelligence Briefs for {len(llm_prompts)} rows...")
llm_outputs = llm.generate(llm_prompts, llm_sampling)

expanded_briefs = []
thinking_logs = []

for idx, output in enumerate(llm_outputs):
    current_index = indices_to_process[idx]

    if not llm_prompts[idx]:
        expanded_briefs.append("")
        thinking_logs.append({"source_index": current_index, "llm_thinking": ""})
    else:
        raw_output_text = output.outputs[0].text.strip()
        combined_text = f"ORIGINAL NARRATIVE: {texts_to_process[idx]}\nEXPANDED BRIEF: {raw_output_text}"
        expanded_briefs.append(combined_text)

        # Standardized tracking inside non-model-specific tags
        think_match = re.search(
            r"<think>(.*?)</think>", raw_output_text, flags=re.DOTALL | re.IGNORECASE
        )
        extracted_thought = think_match.group(1).strip() if think_match else ""

        thinking_logs.append(
            {
                "source_index": current_index,
                "llm_thinking": extracted_thought,
                "llm_brief": raw_output_text,
            }
        )

df_thinking = pd.DataFrame(thinking_logs)

# ==============================================================================
# STEP 1.5: VRAM FLUSH
# ==============================================================================
print("\n--- FLUSHING VRAM ---")
del llm
del tokenizer
gc.collect()
torch.cuda.empty_cache()
print("VRAM cleared. Ready for Step 2.\n")

# ==============================================================================
# STEP 2: THE EXTRACTOR (NuExtract 1.5)
# ==============================================================================

nu_id = "numind/NuExtract-1.5"
print(f"--- STEP 2: Loading {nu_id} ---")

nu_llm = LLM(
    model=nu_id,
    dtype="bfloat16",
    tensor_parallel_size=1,
    max_model_len=8192,
    gpu_memory_utilization=0.90,
    enforce_eager=False,
    trust_remote_code=True,
)

# Streamlined Schema: Target attributes omitted since exact positions are confirmed
nu_template_dict = {
    "weapons_used": [
        {
            "weapon_class": "",  # e.g., "Air-to-Surface / Smart Bomb", "Cruise Missile"
            "estimated_quantity": 0,
        }
    ],
    "ground_impact_status": "",
    "structural_impact": "",  # Captures exact string from Step A
}

json_template = json.dumps(nu_template_dict, indent=2)

nu_prompts = []
for brief in expanded_briefs:
    if not brief:
        nu_prompts.append("")
    else:
        nu_prompt = f"<|input|>\n### Template:\n{json_template}\n### Text:\n{brief}\n\n<|predict|>"
        nu_prompts.append(nu_prompt)

nu_sampling = SamplingParams(temperature=0.0, max_tokens=2000)

print("Extracting strict JSON munitions parameters...")
nu_outputs = nu_llm.generate(nu_prompts, nu_sampling)

# ==============================================================================
# STEP 3: BULLETPROOF JSON PARSING (STANDARDIZED & INDEPENDENT)
# ==============================================================================
print("\n--- STEP 3: PARSING FINAL JSON ---")
import json
import re

all_results = []

# Fallback mechanism if json_repair happens to be available, otherwise use standard json
try:
    import json_repair
except ImportError:
    json_repair = None


def extract_json_block(text):
    """Safely extracts and parses the true JSON block from LLM output,

    handling conversational text, tags, or formatting duplicates.
    """
    if not isinstance(text, str) or not text.strip():
        return None

    # Strategy 1: Check for standard NuExtract target blocks between tags
    predict_matches = re.findall(
        r"<\|predict\|>(.*?)(?:<\|end_predict\|>|$)", text, re.DOTALL | re.IGNORECASE
    )
    if predict_matches:
        for match in predict_matches:
            clean = match.strip()
            if "{" in clean and "}" in clean:
                s_idx = clean.find("{")
                e_idx = clean.rfind("}")
                candidate = clean[s_idx : e_idx + 1]
                try:
                    return (
                        json_repair.loads(candidate)
                        if json_repair
                        else json.loads(candidate)
                    )
                except Exception:
                    continue

    # Strategy 2: Scan for any valid isolated brace structures sequentially
    start_positions = [m.start() for m in re.finditer(r"\{", text)]
    end_positions = [m.start() for m in re.finditer(r"\}", text)]

    for start in start_positions:
        for end in sorted(end_positions):
            if end > start:
                candidate = text[start : end + 1]
                try:
                    return (
                        json_repair.loads(candidate)
                        if json_repair
                        else json.loads(candidate)
                    )
                except Exception:
                    continue
    return None


# Execute extraction over the raw text outputs stored in memory
for idx, output in enumerate(nu_outputs):
    current_index = indices_to_process[idx]

    if not nu_prompts[idx]:
        continue

    raw_response = output.outputs[0].text.strip()
    parsed_dict = extract_json_block(raw_response)

    if parsed_dict and isinstance(parsed_dict, dict):
        parsed_dict["status"] = "SUCCESS"
        parsed_dict["source_index"] = current_index
        all_results.append(parsed_dict)

print(
    f"Successfully parsed strict JSON structures for {len(all_results)} out of {len(nu_outputs)} rows."
)

# ==============================================================================
# STEP 4: PANDAS FORMATTING & DETERMINISTIC PHYSICS ENFORCER
# ==============================================================================
print("\n--- STEP 4: ENFORCING SINGLE IMPACT RADIUS & MERGING ---")

df_results = pd.DataFrame(all_results)

# Build an empty structured skeleton if all extractions failed, otherwise normalize
if df_results.empty or "source_index" not in df_results.columns:
    print("WARNING: Extraction pool is empty. Proceeding with blank entries.")
    df_expanded = pd.DataFrame(
        columns=[
            "source_index",
            "weapons_used",
            "structural_impact",
            "max_crater_radius_m",
            "total_damage_footprint_m",
        ]
    )
else:
    df_expanded = df_results.copy()
    new_columns = {col: col.split(".")[-1] for col in df_expanded.columns if "." in col}
    df_expanded = df_expanded.rename(columns=new_columns)


def calculate_salvo_impact(row):
    weapons_list = row.get("weapons_used", [])
    impact = str(row.get("structural_impact", "")).lower()

    radii_chart = {
        "small kamikaze": 5,
        "loitering": 15,
        "shahed": 15,
        "rocket": 12,
        "artillery": 20,
        "mortar": 15,
        "air-to-surface": 40,
        "smart bomb": 40,
        "jdam": 45,
        "cruise": 35,
        "ballistic": 60,
    }

    max_single_radius = 0
    total_munitions = 0

    if isinstance(weapons_list, list):
        for w in weapons_list:
            if not isinstance(w, dict):
                continue
            w_class = str(w.get("weapon_class", "")).lower()
            qty = w.get("estimated_quantity", 1)

            try:
                qty = int(qty) if qty is not None else 1
            except ValueError:
                qty = 1

            total_munitions += qty

            for key, val in radii_chart.items():
                if key in w_class:
                    if val > max_single_radius:
                        max_single_radius = val
                    break

    if max_single_radius == 0:
        max_single_radius = 18
    if total_munitions == 0:
        total_munitions = 1

    multiplier = 1.0
    if "total annihilation" in impact:
        multiplier += 0.5
    elif "targeted" in impact:
        multiplier -= 0.5
    elif "minor" in impact:
        multiplier -= 0.7

    final_max_crater = int(max_single_radius * max(0.1, multiplier))
    total_facility_footprint = final_max_crater + (total_munitions * 10)

    return final_max_crater, total_facility_footprint


# Safely map calculated tuple elements using list comprehension to avoid assignment errors
if not df_expanded.empty and "weapons_used" in df_expanded.columns:
    impact_metrics = df_expanded.apply(calculate_salvo_impact, axis=1).tolist()
    df_expanded["max_crater_radius_m"] = [m[0] for m in impact_metrics]
    df_expanded["total_damage_footprint_m"] = [m[1] for m in impact_metrics]
else:
    df_expanded["max_crater_radius_m"] = None
    df_expanded["total_damage_footprint_m"] = None

# Left join ensures all original records remain completely untouched
df_final = df_exact.merge(
    df_expanded, left_index=True, right_on="source_index", how="left"
)
df_final = df_final.loc[:, ~df_final.columns.duplicated()]

if not df_thinking.empty:
    df_final = df_final.merge(df_thinking, on="source_index", how="left")

print(
    f"Pipeline complete! Successfully integrated all data points into {len(df_final)} exact coordinate rows."
)

export_filename = os.path.join(base_folder, "final_extracted_airstrike_events.csv")
df_final.to_csv(export_filename, index=False, encoding="utf-8")
print(f"[SUCCESS] Core analysis output generated at: {export_filename}")
```

```python
from sklearn.cluster import DBSCAN
import numpy as np

# 1. Filter for exact coordinate sites
df_exact = df_clean[df_clean['coord_type'] == 'exact'].copy()

# 2. Convert Lat/Lon to Radians for the Haversine formula
coords = np.radians(df_exact[['Calculated_Lat', 'Calculated_Lon']].values)

# 3. Define proximity (e.g., group strikes within 2 kilometers of each other)
# Earth's radius is ~6371 km. 2km / 6371km = 0.0003138 radians
kms_per_radian = 6371.0
epsilon = 2.0 / kms_per_radian

# 4. Run the spatial clustering algorithm
db = DBSCAN(eps=epsilon, min_samples=3, metric='haversine').fit(coords)
df_exact['target_cluster_id'] = db.labels_

# Summary of target zones found
unique_clusters = len(set(db.labels_)) - (1 if -1 in db.labels_ else 0)
print(f"Identified {unique_clusters} high-density tactical strike zones among exact coordinates.")
print(f"Flagged {list(db.labels_).count(-1)} strikes as isolated/scattered events.")
```

```python
# REVISED: Tight, physics-defending modifiers
SITE_MULTIPLIERS = {
    # Sprawling industrial/aviation hubs (slight bump for secondary fire/shrapnel)
    "military : airbase": 1.10,      # 50m baseline becomes 55m max
    "civilian : industrial": 1.05,  # 50m baseline becomes 52m max
    "civilian : port": 1.05,
    "energy : oil": 1.10,           # High secondary fire risk
    "energy : gas": 1.10,

    # Standard or hardened concrete targets (Zero modification / direct shockwave)
    "military : hq": 1.00,          # Keeps baseline exactly at 50m
    "military : base": 1.00,
    "military": 1.00,

    # Soft, unreinforced structural propagation
    "civilian": 1.05,
    "unknown": 1.00
}
import numpy as np
import pandas as pd

print("--- EXECUTING OPTIMIZED TARGET CORRIDOR RADIUS MAPPING ---")

# Isolate the clean 351 exact pinpoint coordinate locations
df_exact_radii = df_clean[df_clean["coord_type"] == "exact"].copy()

# 1. Establish the explicit, capped weapon baseline parameters
BASE_WEAPON_RADII = {
    "missile": 50,      # Capped exactly at 50 meters per your specification
    "drone": 15,        # Loitering / Kamikaze baseline
    "rocket": 12,       # Standard tactical artillery rocket
    "mortar": 8,        # Light infantry mortar
    "air defense": 4,   # Interception fragments / falling debris
    "unknown": 15,      # Conservative fallback baseline
}

def determine_optimized_radius(row):
    # Step A: Identify the core weapon category from event_type
    event_str = str(row.get("event_type", "")).lower()

    selected_class = "unknown"
    if "missile" in event_str:
        selected_class = "missile"
    elif "drone" in event_str:
        selected_class = "drone"
    elif "rocket" in event_str:
        selected_class = "rocket"
    elif "mortar" in event_str:
        selected_class = "mortar"
    elif "defense" in event_str or "activity" in event_str:
        selected_class = "air defense"

    base_radius = BASE_WEAPON_RADII[selected_class]

    # Step B: Scan hierarchy for the best available target structure profile
    site_stype = str(row.get("siteStype", "")).lower().strip()
    site_type = str(row.get("site_type", "")).lower().strip()

    # Determine the structural key lookup precedence
    if site_stype and site_stype != "unspecified" and site_stype in SITE_MULTIPLIERS:
        multiplier_key = site_stype
    elif site_type and site_type != "unknown" and site_type in SITE_MULTIPLIERS:
        multiplier_key = site_type
    else:
        multiplier_key = "unknown"

    modifier = SITE_MULTIPLIERS[multiplier_key]

    # Step C: Finalize physics boundaries
    final_radius_m = int(base_radius * modifier)
    impact_area_sqm = int(np.pi * (final_radius_m ** 2))

    return pd.Series([selected_class, multiplier_key, final_radius_m, impact_area_sqm])

# Apply calculations across the dataset vectors
df_exact_radii[[
    "derived_weapon_class",
    "radius_lookup_source",
    "final_calculated_radius_m",
    "impact_footprint_sqm"
]] = df_exact_radii.apply(determine_optimized_radius, axis=1)

# ==============================================================================
# VIEW FINAL AUDIT QUALITY ASSURANCE MATRIX
# ==============================================================================
print("\n### FINAL EXTRACTION BOUNDARY PROFILE METRICS ###")
summary_matrix = (
    df_exact_radii.groupby("derived_weapon_class")
    .agg(
        Events_Processed=("event_id", "count"),
        Max_Assigned_Radius_M=("final_calculated_radius_m", "max"),
        Min_Assigned_Radius_M=("final_calculated_radius_m", "min"),
        Avg_Impact_Area_SqM=("impact_footprint_sqm", "mean"),
    )
    .sort_values(by="Events_Processed", ascending=False)
)
print(summary_matrix)

# Save off final structured mapping
output_path = os.path.join(base_folder,"exact_coordinates_retaliation_radii.csv")
df_exact_radii.to_csv(output_path, index=False)
print(f"\n[SUCCESS] Matrix processed and written to: '{output_path}'")
```

```python
import pandas as pd

def build_event_title(row):
    # Helper function to clean text and filter out "UNK"
    def clean_text(val):
        if pd.isna(val):
            return ""
        val_str = str(val).strip()
        if val_str.upper() in ["UNK", "UNKNOWN", ""]:
            return ""
        return val_str

    # 1. Actor and Event
    actor = clean_text(row['actor'])
    event = clean_text(row['event_type']) or 'Strike'

    # Combine actor and event (if no actor, it just starts with the event)
    action_str = f"{actor} {event}".strip()

    # 2. Subject and Site Type
    subject = clean_text(row['subject'])
    site = clean_text(row['site_type'])

    target_str = ""
    if not subject and site:
        target_str = f"{site} Site" # e.g., "Civilian Site"
    elif subject and not site:
        target_str = subject
    elif subject and site:
        target_str = f"{subject} ({site})"

    # 3. Location
    loc_parts = [clean_text(x) for x in [row['city'], row['province'], row['country']]]
    loc_parts = [x for x in loc_parts if x] # Keep only non-empty strings
    location_str = ", ".join(loc_parts)

    # 4. Build the final sentence dynamically
    parts = [action_str]

    if target_str:
        parts.append(f"on {target_str}")

    if location_str:
        parts.append(f"in {location_str}")

    # Combine all parts and capitalize for title case
    title = " ".join(parts)

    return title.title()

# Apply the function
df_retaliation['event_full_title'] = df_retaliation.apply(build_event_title, axis=1)
```

```python
import pandas as pd
import numpy as np
import os

print("--- STARTING FAIL-SAFE DATAFRAME UNIFICATION ---")
base_folder ='/content/drive/MyDrive/Iran Israel War/extracted_map_layers'
retaliation_path = os.path.join(base_folder, "exact_coordinates_retaliation_radii.csv")

# ==============================================================================
# 1. Load the dataframes (Preserving your working copies)
# df_retaliation = pd.read_csv(retaliation_path)
# df_airstrikes = pd.read_csv(os.path.join(base_folder, "final_extracted_airstrike_events.csv"))
df_airstrikes['Calculated_Lat']=df_airstrikes['Latitude']
df_airstrikes['Calculated_Lon']=df_airstrikes['Longitude']
df_retaliation['Calculated_Lat']=df_retaliation['latitude']
df_retaliation['Calculated_Lon']=df_retaliation['longitude']

print("--- EXECUTING CLEAN-SLATE DATAFRAME UNIFICATION ---")

# 1. Take safe copies of your source dataframes to protect them
df1_source = df_airstrikes.copy()
df2_source = df_retaliation.copy()
print('event_full_title' in df1_source.columns)
print('event_full_title' in df2_source.columns)
print(f"Loaded Phase 1 (Airstrikes): {len(df1_source)} rows")
print(f"Loaded Phase 2 (Retaliation): {len(df2_source)} rows")


def extract_standardized_phase(df_source, phase_name):
    """Builds a completely clean, uncorrupted DataFrame column-by-column

    safely resolving duplicate names and column mismatches.
    """
    df_new = pd.DataFrame()

    # 1. Inject tracking metadata tags
    df_new['campaign_phase'] = [phase_name] * len(df_source)
    df_new['original_source_row'] = list(df_source.index)

    # 2. Map standard text & categorical columns
    standard_cols = ['event_id', 'strikedate', 'actor', 'side', 'event_type', 'site_type', 'siteStype', 'country','event_full_title']
    for col in standard_cols:
        if col in df_source.columns:
            col_data = df_source[col]
            # Safety: If the column is duplicated in source, col_data is a DataFrame. Take the first one.
            if isinstance(col_data, pd.DataFrame):
                df_new[col] = col_data.iloc[:, 0].values
            else:
                df_new[col] = col_data.values
        else:
            print(col)
            df_new[col] = np.nan

    # 3. Coordinate Resolution: Prioritize precise 'Calculated' coordinates over broad ones
    for target_coord, source_candidates in [('latitude', ['Calculated_Lat', 'latitude']),
                                            ('longitude', ['Calculated_Lon', 'longitude'])]:
        found = False
        for cand in source_candidates:
            if cand in df_source.columns:
                col_data = df_source[cand]
                if isinstance(col_data, pd.DataFrame):
                    df_new[target_coord] = col_data.iloc[:, 0].values
                else:
                    df_new[target_coord] = col_data.values
                found = True
                break
        if not found:
            df_new[target_coord] = np.nan

    # 4. Weapon Blast Radius Resolution
    radius_candidates = ['final_calculated_radius_m', 'max_crater_radius_m', 'calculated_radius_m']
    found_radius = False
    for cand in radius_candidates:
        if cand in df_source.columns:
            col_data = df_source[cand]
            if isinstance(col_data, pd.DataFrame):
                df_new['final_calculated_radius_m'] = col_data.iloc[:, 0].values
            else:
                df_new['final_calculated_radius_m'] = col_data.values
            found_radius = True
            break
    if not found_radius:
        df_new['final_calculated_radius_m'] = np.nan

    # 5. Blast Impact Area Footprint Resolution (Ensure identical math dimensions)
    if 'impact_footprint_sqm' in df_source.columns:
        col_data = df_source['impact_footprint_sqm']
        if isinstance(col_data, pd.DataFrame):
            df_new['impact_footprint_sqm'] = col_data.iloc[:, 0].values
        else:
            df_new['impact_footprint_sqm'] = col_data.values
    elif 'total_damage_footprint_m' in df_source.columns:
        # Map Phase 1 linear metrics safely to square meters area: pi * r^2
        col_data = df_source['total_damage_footprint_m']
        if isinstance(col_data, pd.DataFrame):
            r_vals = pd.to_numeric(col_data.iloc[:, 0], errors='coerce').fillna(0).values
        else:
            r_vals = pd.to_numeric(col_data, errors='coerce').fillna(0).values
        df_new['impact_footprint_sqm'] = (np.pi * (r_vals ** 2)).astype(int)
    else:
        # Fallback: compute from the calculated radius column directly
        r_vals = pd.to_numeric(df_new['final_calculated_radius_m'], errors='coerce').fillna(0).values
        df_new['impact_footprint_sqm'] = (np.pi * (r_vals ** 2)).astype(int)

    return df_new


# Construct clean, uncorrupted sets for both phases
df_airstrikes_final = extract_standardized_phase(df1_source, 'Initial Airstrike Wave')
df_retaliation_final = extract_standardized_phase(df2_source, 'Iranian Retaliation Wave')

print("\nVerifying Column Structural Health:")
print(f"  → Airstrikes unique columns? {df_airstrikes_final.columns.is_unique} (Count: {len(df_airstrikes_final.columns)})")
print(f"  → Retaliation unique columns? {df_retaliation_final.columns.is_unique} (Count: {len(df_retaliation_final.columns)})")

# Stack cleanly without index remapping failures
df_master_campaign = pd.concat([df_airstrikes_final, df_retaliation_final], axis=0, ignore_index=True)

# Format dates and sort chronologically for a single linear unified war log
df_master_campaign['strikedate'] = pd.to_datetime(df_master_campaign['strikedate'], errors='coerce')
df_master_campaign = df_master_campaign.sort_values(by='strikedate').reset_index(drop=True)

print(f"\n[SUCCESS] Matrix unified! Combined Log Rows: {len(df_master_campaign)}")
print(df_master_campaign['campaign_phase'].value_counts())

# Export master integrated asset
master_output_path = os.path.join(base_folder, "master_unified_campaign_log.csv")
df_master_campaign.to_csv(master_output_path, index=False)
print(f"Master file successfully compiled and written to: {master_output_path}")
```
